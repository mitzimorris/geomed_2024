# Spatial Data Analysis in Python and R

## Overview
 
Areal data consists of a finite set of regions with well-defined boundaries.
Geographic information systems (GIS) encode areal boundaries as a set of bounding polygons.
To represent other features, e.g. area name and demographic together with GIS data
we use packages which extend tabular dataframes to GIS data

- Python: [`GeoPandas`](https://geopandas.org/en/stable/index.html) - add support for geographic data to pandas objects. 
- R: [`sf`](https://r-spatial.github.io/sf/) Simple Features for R - "spatial analysis simplified"

#### Region outlines - prefer low-resolution 

Computing all neighbors for a map with many regions is computationally expensive;
all polygons making up one region must be checked against all polygons of other regions
for overlapping, shared, or very close boundaries or points.
For this reason it is advisable to work with maps with a very coarse resolution
as each area is described by fewer polygons, thus requiring less compute.

#### Neighbor graphs

By defining a neighbor relationship between area regions, we take the information from a GIS dataframe
and convert it to a graph, where the areas are the nodes and edges denote the neighbor relationship.
To compute neighbor graphs we use the following packages

- Python: [`libpysal`](https://pysal.org/libpysal/) the Python Spatial Analysis Library Core
- R: [`spdep`](https://r-spatial.r-universe.dev/spdep) Spatial Dependence: Weighting Schemes, Statistics

There are many ways to define contiguity, of these two straightforward metrics
are *Rook* and *Queen* - regions which share bounding line are neighbors,
and regions which share a bounding line or vertex are neighbors, respectively.
The first sections this notebook show how to create, plot, and analyze
spatial dataframes and neighbor graphs.

#### When geographic boundaries and conceptual boundaries don't line up

Sometimes the contiguity graph doesn't line up
with the conceptual neighbor graph imposed according to some organizational schema.
For example, the map may be drawn such that although a river separates one region from another,
if the map is drawn such that they share a border somewhere at the bottom of the river.
However if the river effectively prevents contact between inhabitants of the two regions,
then for some analysis, they should not be considered neighbors and the graph must be modified accordingly.
The later sections of this notebook show you how to do this.

### Spatial Data manipulation in R and Python

Geographic information systems (GIS) data is any item which has a geographic location, either a single point or a set of bounding polygons.
Cartographic data (maps) are encoded as a set of records, one per map region.  The [shapefile format](https://en.wikipedia.org/wiki/Shapefile)
is an open specification used to insure interoperability among GIS software packages.  When items in a dataset contain location labels,
it is necessary to obtain a set of shapefiles for the corresponding map.
To compute with GIS data, we need packages which can do the geographic math.

- In Python the [GeoPandas](https://geopandas.org/en/stable/) package extends the datatypes used by pandas to allow spatial operations on geometric types,
and the [`libpysal`](https://pysal.org/libpysal/) package analyzes spatial relationships between locations.
- In R the equivalent operations are provided by the [`sf`](https://r-spatial.github.io/sf/) and the [`spdep`](https://r-spatial.github.io/spdep/) package.

To visualize GIS data, we need to be able to draw maps and spatial graphs.

- In R, ggplot2 can produce beautiful maps for GIS data.

## From spatial dataframes to graphs

The neighbor graph is constructed from a geodataframe by
analysis of the bounding polygons in the `geometry` column.
Both libpysal and spdep provide functions to visualize neighbor graph overlaid on the map.
The neighbor graph objects merely index entries in the geodataframe.
Therefore plotting requires extracting the centroids of each regions
from the geodataframe in order to align the graph to the map.
Because the default indexing scheme is row index, it is critical
that the dataframe and the neighbor graph remain in sync;
reordering or otherwise modifying the geodataframe will completely
scramble the graph and the resulting plot will look nothing like
a proper network.

````python
nyc_nbs = libpysal.weights.Rook.from_dataframe(nyc_geodata, geom_col='geometry')
plot_spatial_weights(nyc_nbs, nyc_geodata)
````

We examine the set of neighbors for the first 3 nodes in the graph.

````python
print('neighbors', list(nyc_nbs.neighbors.items())[:3])
print('cardinalities', list(nyc_nbs.cardinalities.values())[:3])
````

### Connected and Disconnected Graphs: Components and Islands

A graph component is a set of nodes for which there is a path from any node in the component
to every other node in that set.
A fully connected graph is a graph over a single component.
As the map of NYC shows (above), the graph is not fully connected, i.e.,
it is impossible to find a path from Staten Island to the Bronx,
and there are 3 regions which have no neighbors - these are islands,
(also called singleton nodes).

The `libpysal` and `spdep` APIs both provide a series of methods and attributes for computing
properties of the neighbor graph.

````python
print(f'max number of neighbors per node: {nyc_nbs.max_neighbors}')
print(f'mean number of neighbors per node: {nyc_nbs.mean_neighbors}')
print(f'islands? {nyc_nbs.islands}')
print(f'number of components: {nyc_nbs.n_components}')
unique, counts = np.unique(nyc_nbs.component_labels, return_counts=True)
print('nodes per component')
for u, c in zip(unique, counts):
    print(f'component {u}: {c}')
````

### Rook vs Queen - does it make a difference?

Consider the non-edge squares on a chessboard.
By the Rook metric each square has 4 neighbors

How does this play out when the map is not as regular as a chessboard?

````python
nyc_nbs_queen = libpysal.weights.Queen.from_dataframe(nyc_geodata, geom_col='geometry')
print(f'queen max number of neighbors per node: {nyc_nbs_queen.max_neighbors}')
print(f'rook max number of neighbors per node: {nyc_nbs.max_neighbors}')
print(f'queen mean number of neighbors per node: {nyc_nbs_queen.mean_neighbors}')
print(f'rook mean number of neighbors per node: {nyc_nbs.mean_neighbors}')
print(f'queen islands? {nyc_nbs_queen.islands}')
print(f'rook islands? {nyc_nbs.islands}')
print(f'queen number of components: {nyc_nbs_queen.n_components}')
print(f'rook number of components: {nyc_nbs.n_components}')
````

## Editing Neighbor Graph Objects

Oftentimes the geospatial structure obtained from the map boundaries may conflict with the logical spatial structure.
Two regions may be map-wise adjacent but if the boundary between them is a 
river or road which prevents contact or mixing between the two regions, then they might not be considered "neighbors".
The converse is also problematic: the geo-coordinates of two regions may not share a bounding line or point but should be considered adjacent for the purposes of the analysis.

The NYC neighbor graph is problematic in both of these two ways.
The first problem - map-wise adjacent but not logically adjacent - is that
there are multiple edges between Manhattan, on one side of the East River, and Brooklyn, and Queens, on the other side.
Perhaps this is due to administrative boundaries to account for bridges, tunnels, and subways, or else it could simply be a map artifact.
For the analysis of pedestrian traffic, these regions should not be connected.
The second problem arises if we want to use a single ICAR component to account for spatial variance
which requires a fully connected neighbor graph.

To modify the graph so that it reflects the desired logical structure requires
editing the libpysal `W` and spdep `nb` objects and to do this
it is necessary to understand their respective APIs.

### The libpysal `W` object API

This object contains 3 member dictionaries, plus several attributes.
The key member variables are:

* `W.neighbors`: A dict mapping each region ID to a list of neighboring region IDs.

* `W.weights`: A dict mapping each region ID to a list of corresponding weights (aligned with neighbors).
These quantify the strength of the relationship (often adjacency or proximity) between different geographic areas.
Because the Rook metric computes a simple binary relationship, the value of all weights is $1.0$.

* `W.cardinalities`: A dict mapping each region ID to its number of neighbors.

If weights object is created without specifying the region IDS, as we did above:

```python
nyc_nbs = libpysal.weights.Rook.from_dataframe(nyc_geodata, geom_col='geometry')
```

The positional row indices of the geodataframe are used as ids.

````python
print('neighbors', list(nyc_nbs.neighbors.items())[:3])
print('cardinalities', list(nyc_nbs.cardinalities.values())[:3])
````

Although the `neighbors`, `weights` and `cardinalities` dictionaries are mutable,
the attribute `n_components` is computed on object instantiation,
therefore after any edits to the `neighbors` and `weights` dicts,
it is necessary to instantiate a new object using the edited `neighbors` and `weights` dictionaries.
The `cardinalities` list will be recomputed, along with `n_components` and `component_labels`.

## Deleting edges from the neighbor graph

To delete an edge between nodes `i` and `j` in the libpysal `W` object
requires updating the `neighbors` and `weights` entries
for both `i` and `j` as well as decrementing the cardinality of each.
If this edit disconnects the graph or subgraph, it is necessary to
instantiate a new `W` object from the `neighbors` and `weights` list
in order to recompute the `n_components` and `component_labels` attributes.

````python
def delete_symmetric_edge(w, i, j):
    """
    Remove the neighbor relationship between i and j in a libpysal.weights.W object.
    Parameters:
    - w: libpysal.weights.W object
    - i: int or str, node identifier
    - j: int or str, node identifier
    """
    # If j is in i's neighbor list, remove it (and corresponding weight)
    if j in w.neighbors[i]:
        idx = w.neighbors[i].index(j)
        w.neighbors[i].pop(idx)
        w.weights[i].pop(idx)
        w.cardinalities[i] = len(w.neighbors[i])
    
    # If i is in j's neighbor list, remove it (and corresponding weight)
    if i in w.neighbors[j]:
        idx = w.neighbors[j].index(i)
        w.neighbors[j].pop(idx)
        w.weights[j].pop(idx)
        w.cardinalities[j] = len(w.neighbors[j])		     
````

### Example: Manhattan as an island

Instead of targeting the spurious connections directly via the node indices,
we instead use the `BoroName` column of the `nyc_geodata` geodataframe
to get per-borough lists of node ids and use them to identify which edges
to drop from the graph.

The geodataframe `nyc_geodata` contains column `BoroName`.
We use this to get the list of row indices for each of the boroughs
Manhattan, Brooklyn, and Queens.

````python
manhattan_indices = nyc_geodata[nyc_geodata['BoroName'] == 'Manhattan'].index
brooklyn_indices = nyc_geodata[nyc_geodata['BoroName'] == 'Brooklyn'].index
queens_indices = nyc_geodata[nyc_geodata['BoroName'] == 'Queens'].index
brooklyn_and_queens = brooklyn_indices.append(queens_indices)

print('Manhattan census tracts: ', len(manhattan_indices))
print('Brooklyn and Queens census tracts: ', len(brooklyn_and_queens))
````

Because these lists have 274 and 13,884 entries respectively,
we use set operations for efficient computation, instead of
repeated calls to `remove_symmetric_edge`.

````python
def clean_nbs_symmetric(w, target_indices, remove_indices):
    """
    Efficiently edit a neighbor graph by removing specified edges symmetrically.
    Parameters:
    - w: libpysal.weights.W object
    - target_indices: list of node identifiers
    - remove_indices: list of node identifiers
    """
    remove_set = set(remove_indices)  # Convert to set for O(1) lookups
    for node in target_indices:
        original_neighbors = w.neighbors[node]
        original_weights = w.weights[node]
        
        # Filter out neighbors to remove
        mask = [neighbor not in remove_set for neighbor in original_neighbors]
        w.neighbors[node] = [neighbor for neighbor, keep in zip(original_neighbors, mask) if keep]
        w.weights[node] = [weight for weight, keep in zip(original_weights, mask) if keep]
        w.cardinalities[node] = len(w.neighbors[node])
        
        # Remove the target node from each of the removed neighbors
        neighbors_to_remove = remove_set.intersection(original_neighbors)
        for neighbor in neighbors_to_remove:
            if node in w.neighbors[neighbor]:
                idx = w.neighbors[neighbor].index(node)
                w.neighbors[neighbor].pop(idx)
                w.weights[neighbor].pop(idx)
                w.cardinalities[neighbor] = len(w.neighbors[neighbor])
````

After calling this function, we replot the map to check our work.

````python
clean_nbs_symmetric(nyc_nbs, manhattan_indices, brooklyn_and_queens)
plot_spatial_weights(nyc_nbs, nyc_geodata)
````

Further checks on the Python object require instantiating a new `W` object in order
to recompute its attributes.

````python
nyc_nbs_clean = libpysal.weights.W(nyc_nbs.neighbors, nyc_nbs.weights)

print(f'is symmetric? {len(nyc_nbs_clean.asymmetries) == 0}')
print(f'number of components: {nyc_nbs_clean.n_components}')
unique, counts = np.unique(nyc_nbs_clean.component_labels, return_counts=True)
print('component ids ', unique, '\ncomponent size', counts)
````

## Adding edges to the neighbor graph

To add an edge between nodes `i` and `j` in the libpysal `W` object
requires updating the `neighbors` and `weights` entries
for both `i` and `j` as well as incrementing the cardinality of each.
If this edit changes the component structure of the neighbor graph
you must instantiate a new `W` object from the `neighbors` and `weights` list.

````python
def add_symmetric_edge(w, i, j, weight=1.0):
    """
    Add a symmetric neighbor relationship between nodes i and j in a libpysal.weights.W object.

    Parameters:
    - w: libpysal.weights.W object
    - i: int or str, node identifier
    - j: int or str, node identifier
    - weight: float, weight to assign to the edge
    """
    if j not in w.neighbors[i]:
        w.neighbors[i].append(j)
        w.weights[i].append(weight)
        w.cardinalities[i] += 1
    if i not in w.neighbors[j]:
        w.neighbors[j].append(i)
        w.weights[j].append(weight)
        w.cardinalities[j] += 1
````

### Example: New York City as single component

Going the other way, instead of removing connections between Manhattan and Brooklyn and Queens,
we need to connect the other components and islands to the Manhattan / Brooklyn / Queens component.
There are 4 additional components:  the Bronx, Staten Island, the Rockaways, and Roosevelt Island,
and three singleton nodes:  City Island, Broad Channel, and Breezy Point.

Since the graph is not connected, there is no correct way to connect these regions, but we try to find reasonably "close" pairs of regions to connect.

````python
# start from geodataframe
nyc_nbs = libpysal.weights.Rook.from_dataframe(nyc_geodata, geom_col='geometry')

add_symmetric_edge(nyc_nbs, 201, 1339) # Spuyten Duyvel to Inwood
add_symmetric_edge(nyc_nbs, 2010, 429) # Grasmere to Bay Ridge
add_symmetric_edge(nyc_nbs, 1311, 1190) # Roosevelt Island to Lenox Hill
add_symmetric_edge(nyc_nbs, 1310, 1385) # Roosevelt Island to Queensbridge
add_symmetric_edge(nyc_nbs, 328, 196) # City Island to Pelham Bay
add_symmetric_edge(nyc_nbs, 1860, 1862)  # Breezy Point to the Rockaways
add_symmetric_edge(nyc_nbs, 1860, 625)  # Breezy Point to Brighton Beach
add_symmetric_edge(nyc_nbs, 1870, 1903)  # Broad Channel to the Rockaways
add_symmetric_edge(nyc_nbs, 1903, 1858)  # Broad Channel to Howard Beech
````

## Comparison of packages `libpysal` and `spdep`

Both Python and R derive exactly the same neighbor graph from a geodataframe,
however the packages differ in the internal structure and methods available for
a libpysal `W` object and a spdep `nb` object.


| **Aspect**                  | **`libpysal.weights.W` (Python)**                                      | **`spdep::nb` (R)**                                 |
|-----------------------------|-------------------------------------------------------------------------|-----------------------------------------------------|
| **Data Structure**          | Dictionaries (`neighbors`, `weights`, `cardinalities`)                 | List of integer vectors                             |
| **Adding Neighbors**        | Append to lists and update weights/cardinalities                       | Insert into sorted vectors                          |
| **Disconnecting Neighbors** | Remove from `neighbors`, `weights`, and update `cardinalities` for both nodes | Remove from neighbor lists and maintain sorted order for both nodes |
| **Symmetry Handling**       | Manually add or remove both directions                                 | Manually add or remove both directions              |
| **Weight Management**       | Directly handled within the `W` object                                 | Typically managed separately (e.g., `listw` objects)|
| **Sorted Neighbor Lists**   | Not inherently sorted                                                  | Maintained in sorted order                          |
